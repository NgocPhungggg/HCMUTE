```python
#Importing libraries

import numpy as np
import pandas as pd
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression


import warnings 
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("d:/DAMI (Data Mining)/Week 05/CaoThiNgocPhung_21110276/train.csv")
```

```python
df.head(10)
```

Data Cleaning

```python
df.isna().sum().sort_values(ascending = False).head(20)
missing_values = pd.DataFrame({'Total_NaN': df.isna().sum().sort_values(ascending = False).head(20)})
missing_values['NaN_Percent'] = missing_values['Total_NaN']/1460
missing_values
```

```python
for variable in missing_values.index:
    if missing_values.loc[variable]['NaN_Percent'] > 0.15:
        df = df.drop(labels = variable, axis = 1)
```

```python
missing_values = pd.DataFrame({'Total_NaN': df.isna().sum().sort_values(ascending = False)})
missing_values['NaN_Percent'] = missing_values['Total_NaN']/1460
missing_values.head(14)
```

```python
missing_values_df= df[list(missing_values.index)]
missing_values_df.info()
```

```python

#Garage Variables

    #GarageYrBlt:
index_with_null_values_GarageYrBlt = df[df['GarageYrBlt'].isna()][['GarageYrBlt']].index
for index in index_with_null_values_GarageYrBlt:
    df['GarageYrBlt'][index] = df['YearBuilt'][index]
df['GarageYrBlt'] = df['GarageYrBlt'].astype(int)
    #Garage các biến phân loại:
df[['GarageType', 'GarageCond', 'GarageFinish', 'GarageQual']] = df[['GarageType', 'GarageCond', 'GarageFinish', 'GarageQual']].fillna('None')


#Biến tầng hầm
df[['BsmtFinType2','BsmtFinType1','BsmtExposure','BsmtCond','BsmtQual']] = df[['BsmtFinType2','BsmtFinType1','BsmtExposure','BsmtCond','BsmtQual']].fillna('None')


#Miếng dán veneer biến đổi
df[['MasVnrArea']] = df[['MasVnrArea']].fillna(0)


#Biến điện
df[['Electrical']] = df[['Electrical']].fillna('Mix')
```

```python
df.isna().sum().max()
rows, cols = df.shape
print(f'After cleaning the dataset, the dataframe has {cols} columns and {rows} rows')
```

Data Analysis


```python
categorical_variables = df.select_dtypes(include='object').columns
categorical_variables
```

Regression

```python
#Nhận hình nộm cho các biến phân loại
df_dummies = pd.get_dummies(df, columns = categorical_variables, drop_first = True)
df_dummies=df_dummies.drop(labels = 'Id', axis=1)
df_dummies.head()
```

```python
from sklearn.model_selection import train_test_split

X = df_dummies.drop(columns='SalePrice')
y = df_dummies['SalePrice']

X_train, X_test, y_train, y_test = train_test_split(X,y,
                                                    test_size   = 0.2,
                                                    random_state = 1234,
                                                    shuffle      = True)
```

```python
print(f'The shape of X_train is {X_train.shape}')
print(f'The shape of X_test is {X_test.shape}')
print(f'The shape of y_train is {y_train.shape}')
print(f'The shape of y_test is {y_test.shape}')
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train) #StandardScaler.fit_transform() chuẩn hóa dữ liệu, đảm bảo rằng chúng có giá trị trung bình bằng 0 và độ lệch chuẩn bằng 1.
X_test_scaled = scaler.transform(X_test)       #StandardScaler.transform() áp dụng phép chuyển đổi dựa trên số liệu thống kê được tính toán trước đó trong fit_transform

#Để có được một df với các cột tên, hãy làm như sau:
X_train_scaled_df = pd.DataFrame(X_train_scaled) #Chúng tôi nhận được một df không có cột tên
X_train_scaled_df.columns = list(df_dummies.columns.drop('SalePrice')) #Chúng tôi thêm tên cột
```

```python
model = LinearRegression()
model.fit(X_train_scaled_df, y_train)
```

```python
#Kiểm tra dự đoán
pred = model.predict(X_train_scaled_df)
pred = pred.flatten()
pred[:10]
```

```python
# Lỗi từ mô hình
rmse_ols = mean_squared_error(
            y_true  = y_train,
            y_pred  = pred,
            squared = False
           )
print("")
print(f"The error(rmse) is: {rmse_ols}")
```

```python
r2_ols = model.score(X_train_scaled_df,y_train)
r2_ols
print(f"The R^2 is: {r2_ols}")
```

```python
# Tính MSE
mse_ols = mean_squared_error(y_true  = y_train,  # Giá trị thực
                             y_pred  = pred,      # Giá trị dự đoán
                             squared = True       # squared=True để tính MSE thay vì RMSE
                            )   
print("")
print(f"The error (MSE) is: {mse_ols}")
```
